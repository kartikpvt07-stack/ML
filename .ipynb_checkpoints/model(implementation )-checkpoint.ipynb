```python
import numpy as np
```

```python
import pandas as pd
```

```python
df = pd.read_csv('./final_data_set.csv')
```

```python
features = [
    "Prev_Return",
    "Prev_Range_Pct",
    "Prev_Direction",

    "Asian_Return",
    "Asian_Range_Pct",
    "Asian_Direction",

    "London_Return",
    "London_Range_Pct",
    "London_Direction",

    "Dist_Prev_High",
    "Dist_Prev_Low"
]

X = df[features]

y = df["Target"]
```

```python
n_features = X.shape[1]

W = np.zeros(n_features)

b = 0
```

```python
print(W)
print(b)
```

```python
z = np.dot(X, W) + b
```

```python
print(z)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
probability= sigmoid(z)
```

```python
print(probability)
```

```python
prediction = (probability >= 0.5).astype(int)
```

```python
print(prediction) # all 0.5
```

<html><h1>logL(β)=∑n​[yi​log(pi​)+(1−yi​)log(1−pi​)]</h1></html>


```python
m = len(y)

cost = -(1/m) * np.sum(
    y*np.log(probability)
    +
    (1-y)*np.log(1-probability)
)

print(cost)
```

<p>
  $$dw = \frac{1}{m} X^T (P - Y)$$
</p>
<p>
  $$db = \frac{1}{m} \sum (P - Y)$$
</p>


```python
dw = (1 / m) * np.dot(X.T, (probability - y))

db = (1 / m) * np.sum(probability - y)

print("dw Shape :", dw.shape)
print("db :", db)
```

```python
learning_rate = 0.01
```

```python
W = W - learning_rate * dw

b = b - learning_rate * db
```

```python

print(W)


print(b)
```

```python
z = np.dot(X, W) + b

probability = sigmoid(z)

print(probability)
```

```python
learning_rate = 0.01
epochs = 1000

m = len(y)

cost_history = []

for epoch in range(epochs):

    
    z = np.dot(X, W) + b
    probability = sigmoid(z)

    
    probability = np.clip(probability, 1e-15, 1 - 1e-15)

    
    cost = -(1 / m) * np.sum(
        y * np.log(probability) +
        (1 - y) * np.log(1 - probability)
    )

    cost_history.append(cost)

    
    dw = (1 / m) * np.dot(X.T, (probability - y))
    db = (1 / m) * np.sum(probability - y)

    
    W = W - learning_rate * dw
    b = b - learning_rate * db

   
    # if (epoch + 1) % 100 == 0:
    #     print(f"Epoch {epoch+1} | Cost = {cost:.6f}")
```

```python
z = np.dot(X, W) + b
probability = sigmoid(z)

prediction = (probability >= 0.5).astype(int)

print(prediction[:100])
```
