```python
df = pd.read_csv("/Volumes/Desk_Data/backtesting/BTCUSD-1m-2026-01.csv")
df.columns = [
    "open_time",
    "open",
    "high",
    "low",
    "close",
    "volume",
    "close_time",
    "quote_volume",
    "num_trades",
    "taker_base",
    "taker_quote",
    "ignore"
]
```

```python
df["open_time"] = pd.to_datetime(
    df["open_time"],
    unit="us",
    utc=True
)

df["close_time"] = pd.to_datetime(
    df["close_time"],
    unit="us",
    utc=True
)
```

```python


df["date"] = df["open_time"].dt.date
df["hour"] = df["open_time"].dt.hour

df[["open_time", "date", "hour"]].tail()
```

```python
def get_session(hour):

    if 0 <= hour < 8:
        return "Asia"

    elif 8 <= hour < 13:
        return "London"

    elif 13 <= hour < 21:
        return "NewYork"

    else:
        return "Other"


df["session"] = df["hour"].apply(get_session)
```
